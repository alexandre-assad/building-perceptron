```python
from numpy import random, mean, zeros
from model.perceptron import Perceptron
```

# Generate random dataframes with values and target values  
We'll create :  
- dataframe of 100 observations with two caracteristics : two numbers between 0 & 1  
- the list of 100 booleans if the respective observation has the sum of its two caracteristics superior to 1 


```python
random.seed(4)

dataframe = random.rand(100, 2) 
dataframe_test = random.rand(100, 2) 
target_values = (dataframe[:, 0] + dataframe[:, 1] > 1).astype(int) 
target_values_test = (dataframe_test[:, 0] + dataframe_test[:, 1] > 1).astype(int) 
```

# Fit and predict with Perceptron

```python
perceptron = Perceptron(2)
perceptron.fit(dataframe, target_values)

# Prédictions
y_pred = perceptron.predict(dataframe_test)
accuracy = mean(y_pred == target_values_test)* 100
print(f"Précision : {accuracy:.2f}%")
```
